# Import data from xml sources 

```python
# Import usefull packages
import xml.etree.ElementTree as ET
import pandas as pd
```

```python
# XML path
xml_file = './/Data//apertium-fra-frp.fra-frp.dix'
```

```python
# Define functions to use
def parse_xml(xml_file):
    tree = ET.parse(xml_file)
    root = tree.getroot()
    return root

def extract_data_custom(root, element_name):
    data = []
    for record in root.findall(f'.//{element_name}'):
        row = {}
        for field in record:
            row[field.tag] = field.text
        data.append(row)
    return data

def to_dataframe_to_csv(data):
    df = pd.DataFrame(data)
    df.to_csv('output.csv', sep= ',')
    return df
```

```python
# Functions calls
root = parse_xml(xml_file)
data = extract_data_custom(root, element_name="section//e//p")
df = to_dataframe_to_csv(data)
```
